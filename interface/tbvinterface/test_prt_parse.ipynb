```python
from parse_prt_file import parse_prt_file

conditions = parse_prt_file('/Users/alexandresayal/GitHub/phd-main-exp/prt/interhemisphericLocalizer340v.prt')

# Print the extracted data
for condition, trials in conditions.items():
    print(f"Condition: {condition}")
    for onset, offset in trials:
        print(f"  Onset: {onset}, Offset: {offset}")
```
